#### Tratamento de Exceções em Python

O bloco try/except é usado para tratar erros (exceções) de forma controlada, evitando que o programa seja interrompido de maneira abrupta quando algo inesperado acontece.
O try testa o codigo, se não acontecer um erro nenhum, o programa segue normalmente. Se acontecer, a execução do bloco try é interrompida e o fluxo passa para o except, onde se trata o error ocorrido.

```python
try:
    numero = input("Digite um número inteiro: ")
    numero_int = int(numero)
    resultado = 10/numero_int
    print("Resultado: ", resultado)

except ValueError:
    print("Erro: você não digitou um número válido")

except ZeroDivisionError:
    print("Erro: não é possível dividir por zero")
```
