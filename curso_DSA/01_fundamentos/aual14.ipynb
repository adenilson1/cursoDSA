### Entrada e Sáida de Dados

```python
# Pedindo o nome do usuário
nome_usuario = input("Qual é o seu nome? ")

#Pedindo a idade do usuário 
idade_usuario = int(input("Qual a sua idade? "))

from datetime import date
# Pega o ano corrente na data definida no sistema operacional da maquina
ano_atual = date.today().year

#Processa os dados
ano_nascimento = ano_atual - idade_usuario

# Saída de dados
print(f"\nOlá, {nome_usuario}: Bem-vindo(a).")
print(f"Você tem {idade_usuario} anos e nasceu aproximadamente em {ano_nascimento}")
```
