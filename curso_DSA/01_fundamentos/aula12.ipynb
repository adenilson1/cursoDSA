### Estrutura de Dados - Set (Conjuntos)

```python
# Criando um conjunto
numeros = {1, 2, 3, 4, 2, 3, 5}
print(f"Conunto de números (sem duplicatas): {numeros}")
```

```python
type(numeros)
```

```python
# Adicionando um item
numeros.add(6)
print(f"Após adicionar o valor 6: {numeros}")
```

```python
# Removendo um item
numeros.remove(2)
print(f"Após remover o valor 2: {numeros}")
```

```python
# Operações de conjunto
conjunto_a = {1, 2, 3, 4}
conjunto_b = {3, 4, 5, 6}
```

```python
# União
uniao = conjunto_a.union(conjunto_b)
print(f"A união de A e B: {uniao}")
```

```python
# Interseção
intersecao = conjunto_a.intersection(conjunto_b)
print(f"A interseção de A e B: {intersecao}")
```
