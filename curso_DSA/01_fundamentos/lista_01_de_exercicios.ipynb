#### Exercício 2
Crie duas variáveis, numero1 e numero2, e atribua a elas valores inteiros. Calcule a soma, subtração, multiplicação e divisão dessas variáveis e imprima os resultados

```python
# Solução
numero1 = int(input("Digite o primeiro numero: "))
numero2 = int(input("Digite o segundo numero: "))

soma = numero1+numero2
subtracao = numero1-numero2
multiplicacao = numero1*numero2
divisao = numero1/numero2

print(f"\nA operações basicas da matematica:")
print(f"\nSoma          de {numero1} + {numero2} = {soma}")
print(f"Subtração     de {numero1} - {numero2} = {subtracao}")
print(f"Multiplicação de {numero1} * {numero2} = {multiplicacao}")
print(f"Divisão       de {numero1} / {numero2} = {divisao}")
```

#### Exercício 4
Crie uma variável chamada saldo com o valor 500.50. Em seguida, crie uma variável saque com o valor 200.25. Subtraia o saque do saldo e imprima o saldo final formatado para duas casa decimais.

```python
# Solução
saldo = 500.50
saque = 200.25
saldo -= saque
print(f"O saldo atual: R$ {saldo:.2f}")
```

#### Exercício 5
Declare uma variável boolena chamada tem_carteira_de_motorista e atribua a ela o valor True. Imprima uma mensagem que diga "Pode dirigir" se a variável for verdeira e "Não pode dirigir" caso contrário

```python
# Solução
idade = int(input("Digite sua idada: "))
sabe_ler_escrever = input("Sabe ler e escrever? (S ou N): ")
rg_cpf = input("Tem RG e CPF? (S ou N):  ")


if (idade >= 18 and sabe_ler_escrever.upper() == 'S' and rg_cpf.upper() == 'S'):
    tem_carteira_de_motorista = True
else:
    tem_carteira_de_motorista = False

if tem_carteira_de_motorista == True:
    print("Pode dirigir")
else:
    print("Não pode dirigir")
```
