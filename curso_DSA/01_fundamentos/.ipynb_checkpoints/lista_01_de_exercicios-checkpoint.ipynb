#### Exercício 2
Crie duas variáveis, numero1 e numero2, e atribua a elas valores inteiros. Calcule a soma, subtração, multiplicação e divisão dessas variáveis e imprima os resultados

```python
# Solução
numero1 = int(input("Digite o primeiro numero: "))
numero2 = int(input("Digite o segundo numero: "))

soma = numero1+numero2
subtracao = numero1-numero2
multiplicacao = numero1*numero2
divisao = numero1/numero2

print(f"\nA operações basicas da matematica:")
print(f"\nSoma          de {numero1} + {numero2} = {soma}")
print(f"Subtração     de {numero1} - {numero2} = {subtracao}")
print(f"Multiplicação de {numero1} * {numero2} = {multiplicacao}")
print(f"Divisão       de {numero1} / {numero2} = {divisao}")
```

#### Exercício 4
Crie uma variável chamada saldo com o valor 500.50. Em seguida, crie uma variável saque com o valor 200.25. Subtraia o saque do saldo e imprima o saldo final formatado para duas casa decimais.

#### Exercício 5
Declare uma variável boolena chamada tem_carteira_de_motorista e atribua a ela o valor True. Imprima uma mensagem que diga "Pode dirigir" se a variável for verdeira e "Não pode dirigir" caso contrário

```python
# Solução
idade = int(input("Digite sua idada: "))
sabe_ler_escrever = input("Sabe ler e escrever? (S ou N): ")
rg_cpf = input("Tem RG e CPF? (S ou N):  ")


if (idade >= 18 and sabe_ler_escrever.upper() == 'S' and rg_cpf.upper() == 'S'):
    tem_carteira_de_motorista = True
else:
    tem_carteira_de_motorista = False

if tem_carteira_de_motorista == True:
    print("Pode dirigir")
else:
    print("Não pode dirigir")
```

#### Exercício 7

#### Exercício 8
Crie duas variáveis booleanas:

chovendo = True

guarda_chuva = False

Use operador lógico para verificar se uma pessoa vai se molhar (se está chovendo E ela não tem gurada-chuva)

```python
# Solução
chovendo = True
guarda_chuva = False

if chovendo ==  True and guarda_chuva == True:
    print("Não vai se molhar")
elif chovendo == True and guarda_chuva == False:
    print("Vai se molhar")
else:
    print("Não vai se molhar")
```

#### Exercício 28
Crie uma lista com os seguintes números: [1, 2, 2, 3, 4, 4, 5, 1]. Use um conjunto para remover os números duplicados e imprima o resultado.

```python
# Solução
lista_de_numeros = [1, 2, 2, 3, 4, 4, 5, 1]

#Convertendo a lista para set para eliminar os numeros duplicados
conjunto_de_numeros = set(lista_de_numeros)

print(f"Lista de numeros com duplicatas: {lista_de_numeros}")
print(f"\nConjunto de numeros (sem duplicatas): {conjunto_de_numeros}")
```

#### Exercício 29
Crie dois conjuntos:

- set_a = {1, 2, 3, 4}
- set_b = {3, 4, 5, 6}
  
Encontre e imprima a interseção entre os dois conjuntos (os elementos que estão em ambos).

```python
# Solução
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

intersecao = set_a.intersection(set_b)

print(f"A interseção entre {set_a} e {set_b} é {intersecao}")
```

#### Exercício 30
Escreva um programa que peça ao usuário para digitar sua altura em metros (ex: 1.75) e seu peso em quilogramas (ex: 68.5). Calcule o Índice de Massa Corporal (IMC) usando a fórmula IMC = peso / (altura * altura) e imprima o resultado formatado com duas casas decimais.

```python
# Solução
print("Olá, digite os seus dados: ")
altura = float(input(f"\nSua altura : "))
peso = float(input(f"Seu peso   : "))

imc = peso/(altura**2)

print(f"\nSeu IMC = {imc:.2f}")
```
