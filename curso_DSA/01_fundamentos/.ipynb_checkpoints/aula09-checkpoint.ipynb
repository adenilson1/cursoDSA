### Estrutura de dados - Listas

```python
# Criando uma lista -> []
frutas = ["maçã", "banana", "laranja", "abacaxi"]
print(f"Lista de frutas: {frutas}")
```

```python
type(frutas)
```

```python
# Acessando um item pelo indice
print(f" A primeira fruta é :{frutas[0]}")
print(f" A última fruta é :{frutas[-1]}")
```

```python
# Adicionando um item ao final da lista
frutas.append("uva")
print(f" A Lista após adicionar 'uva' fruta é :{frutas}")
```

```python
# Remover um item
frutas.remove("laranja")
print(f" A Lista após remover 'laranja' :{frutas}")
```

```python
# Modificando um item
frutas[0] = "morango"
print(f" A Lista após modficar o primeiro item :{frutas}")
```

```python
# Podemos imprimir direatemnte 
print(frutas)
```

```python
# Verificando o tamanho da lista
print(f" A Lista tem {len(frutas)} frutas.")
```

```python
# Deletando a lista
del frutas
```

```python
# A lista nao pode mais ser acessada
print(frutas)
```
