### Manipulação de Strings

```python
# Define uma variável do tipo string
frase = "  Aprender Python é muito divertido!  "
```

```python
# Concatenação
nome = "Maria"
saudacao = "Olá, " + nome + "!"
print(saudacao)
```

```python
# Tamanho da string
print(f"Maiúsculas: {frase.upper()}")
print(f"Minúsculas: {frase.lower()}")
```

```python
# Remover espaços em branco do inicio e do fim
frase_sem_espacos = frase.strip()
print(f"Frase sem espaços: '{frase_sem_espacos}'")
```

```python
# Substitui texto
print(f"Substituir 'divertido' por 'legal': {frase_sem_espacos.replace('divertido', 'legal')}")
```

```python
# Fatiamento (Slicing) - Acessando partes de uma string
# O indice em Python começa em 0
print(frase_sem_espacos)
print(f"O primeiro caractere: {frase_sem_espacos[0]}")
print(f"A palavra 'Python' : {frase_sem_espacos[9:15]}") # do indice 9 ate 14
```
