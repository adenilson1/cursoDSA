```python
# Usando o módulo 'math' para funções matematicas
import math
```

```python
# Calcule a raiz quadrada
raiz_quadrada = math.sqrt(25)
print(f"A raiz quadrada de 25 é: {raiz_quadrada}")
```

```python
# Usando o módulo 'random' para gerar números aleatorios
import random
```

```python
# Gera um inteiro ente 1 a 100
numero_aleatorio = random.randint(1, 100)
print(f"Um número aleatório entre 1 e 100: {numero_aleatorio}")
```

```python
# Seleciona aleatoriamente uma cidade da lista
cidade_alaetorio = random.choice(["Rio de Janeiro", "Salvador", "Curitiba"])
print(f"Cidade escolhida aleatoriamente: {cidade_alaetorio}")
```
