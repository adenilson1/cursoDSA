#### Comprehensions (List, Set, Dict e Generator) 

```python
# Criando uma lista de quadrados dos numeros de 0 a 9
quadrados = [x ** 2 for x in range(10)]
print(f"Quadrados de 0 a 9: {quadrados}")
```

```python
# Criando uma lista de números pares de 0 a 20
numeros_pares = [x for x in range(20) if x %2 == 0]
print(f"Números pares de 0 a 20: {numeros_pares}")
```

```python
type(numeros_pares)
```

```python
# Criando um dicionario com numeros e seus quadrados
quadrados_dict = {x : x **2 for x in range(6) }
print(quadrados_dict)
```

```python
type(quadrados_dict)
```

```python
# Conjunto de quadrados (sem repetiçõs)
quadrados_set = {x ** 2 for x in [1 , 2, 2, 3 ,3, 4]}
print(quadrados_set)
```

```python
type(quadrados_set)
```

```python
# Generator expression (não é tupla ainda)
gen = (x ** 2 for x in range(6))
print(f"Generator : {gen}")

# Convertendo em tupla
quadrados_tuple = tuple(gen)
print(f"Tupla criada a parte da conversção do  generator para tupla: {quadrados_tuple}")
```

```python
type(quadrados_tuple)
```
