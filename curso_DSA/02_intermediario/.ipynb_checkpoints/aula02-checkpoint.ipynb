#### Estruturas de Repetição - Loops For - Parte 01

```python
# Loop for
# Define uma lista
frutas = ["maçã", "banana", "cereja"]

# Imprime a mensagem
print("Frutas disponíveis:")

#loop pela lista
for fruta in frutas:
    print(f"- {fruta}")
```

```python
# Criando uma tupla
cores = ("vermelho", "verde", "azul")

#loop for percorrendo a tupla
for cor in cores:
    print(cor)
```

```python
# Criando um dicionário
formacoes_dsa = {"Formação Cientista de Dados":6, "Formação Analista de Dados":4, "Formação Engenheiro de Dados":5}

#loop
for chave, valor in formacoes_dsa.items():
    print(chave, ":", valor)
```

```python
# Função range()
print("\nContagem até 5:")
for numero in range(6):
    print(numero)
```
