#### Estrutura de Repetição - Loop While - Parte 2

```python
# Define a variável
contador = 5

# Imprime a mensagem
print("Contagem regressiva:")

# Loop
while contador > 0:
    print(contador)
    contador -= 1
```

```python
# Define a variável
contador = 0

# Imprime a mensagem
print("Contagem crescente:")

# Loop
while contador < 3:
    print(contador)
    contador += 1
```
