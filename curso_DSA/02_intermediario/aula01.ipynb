#### Tomada de Decisão com Estruturas Condicionais

```python
# Define a variável
nota = 8.5

# checagem e tomada de decisão
if nota >= 7.0:
    print("Aprovado!")
else:
    print("Reprovado.")
```

```python
# Variável
idade = 17

if idade < 18:
    print("Você é menor de idade.")
elif idade >= 18 and idade < 65:
    print("Você é um adulto.")
else:
    print("Você é idoso.")
```
