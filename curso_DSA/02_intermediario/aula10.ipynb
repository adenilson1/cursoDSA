#### Funções Anônimas (Expressão Lambda)

```python
# Uma função que dobra um número
dobro = lambda x: x*2

#Print
print(f"O dobro de 7 é: {dobro(7)}")
```

```python
# Lista de números
numeros = [1, 2, 3, 4, 5]

# Lmabda que retorna o quadrado de cada elemento
quadrados = list(map(lambda x: x ** 2, numeros))

print(quadrados)
```

```python
# Lista de números
numeros = [1, 2, 3, 4, 5, 6]

# Primeiro calculamos os quadrados com map + lambda
quadrados = list(map(lambda x: x ** 2, numeros))

# Agora filtramos apenas os pares com filte + lambda
quadrados_pares = list(filter(lambda x: x % 2 == 0, quadrados))

print("Quadrados: ", quadrados)
print("Quadrados pres: ", quadrados_pares)
```
