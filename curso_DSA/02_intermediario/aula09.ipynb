#### Trabalho com Números Variado de Argumentos em Funções

```python
# Argumentos de tamanho variável (*args)

def dsa_soma_numeros(*args):
    """Soma um números variável de argumentos"""
    total = 0

    for numero in args:
        total += numero
    return total
```

```python
print(f"Soma dos Números: {dsa_soma_numeros(1, 2, 3, 4, 5)}")
```

```python
print(f"Soma dos Números: {dsa_soma_numeros(1, 2, 3)}")
```

```python
print(f"Soma dos Números: {dsa_soma_numeros(10, 400, 0.3, 120)}")
```

```python
# Argumentos de tamanho variável (** kwargs)
def dsa_exibe_info_pessoa(**kwargs):
    """Exibe informçãoes passdas com pares chave-valor"""
    
    print("\nInformaç~eos da Pessoa:\n")

    for chave, valor in kwargs.items():
        print(f"- {chave}: {valor}")
        
```

```python
# Chamando a função
dsa_exibe_info_pessoa(
                      nome = "Carla",
                      profissao = "Engenharia de Dados",
                      hobby = "Leitura"
)
```

```python
# Chamando a função
dsa_exibe_info_pessoa(
                     nome = "Bob",
                     profissao = "Cientista de Dados"
)
```
